We have Hamiltanian 

$H = \hbar \Delta \ket{e}\bra{e} + \frac{\hbar \Omega}{2}(\ket{e}\bra{g}+\ket{g}\bra{e})$

$\ket{e}$ - exited state

$\ket{g}$ - ground state

$\Delta = \omega_0 - \omega$

$\omega_0$ - frequency of field
$\omega$ - frequency of level

```python
e = basis(2,1)
g = basis(2,0)
h = 1  # Planck's constant
D = -1.0  # Detuning
Omega = 1.0  # Rabi frequency
Gamma = 1  # Decay rate

H = h*D*e*e.dag() + h*Omega*(e*g.dag() + g*e.dag())/2

c_ops = np.sqrt(Gamma)*g*e.dag()  # Collapse operator
H
```

```python
rho_ee_experimental = e.dag() * steadystate(H, [c_ops]) * e
steadystate(H,[c_ops])
```

Formula for speed transmition

$R = \frac{\Gamma}{2}\frac{s}{1+s+(\frac{2\Delta}{\Gamma})^2}$

where $s = \frac{2\Omega^2}{\Gamma^2}$

Also probabiliti for consider in exited state is 

$\rho_{ee} = \frac{1}{2}\frac{s}{1+s+(\frac{2\Delta}{\Gamma})^2}$

```python
s = 2*(Omega**2)/(Gamma**2)
rho_ee = (1/2)*(s)/(1+s+(4*D**2)/(Gamma**2))
rho_ee
```
